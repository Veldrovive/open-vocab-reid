```python
import sys
from pathlib import Path

from open_vocab_mot.data.dataset_factory import load_dataset_for_training, UnsupervisedDatasetConfig
from open_vocab_mot.data.video_reid_abc import VideoReIDBatch, collate_video_reid_ds
from torch.utils.data import DataLoader
from torchvision.transforms.v2.functional import to_pil_image
from IPython.display import display
```

```python
config = UnsupervisedDatasetConfig(
    subset_dataset="house_tours",
    use_for_training=True,
    weight=1.0,
    frames_per_video=4,
    people_per_batch=16,
    views_per_person=2,
    num_hard_negatives_per_positive=1,
    warn_skipped_identities=False
)
```

```python
sampler_ds = load_dataset_for_training(
    "unsupervised", config, seed=42, verbose=True
)
```

```python
sampler_loader = DataLoader(sampler_ds, batch_size=None, collate_fn=collate_video_reid_ds)
```

```python
batch_iter = iter(sampler_loader)
```

```python
batch: VideoReIDBatch = next(batch_iter)
print(f"Batch Size (Frames): {len(batch.frame_tensors)}")
batch
```

```python
for i in range(len(batch.frame_tensors)):
    print(f"Identity: {batch.identity_ids[i].item()} | Sequence: {batch.sequence_ids[i].item()}")
    display(to_pil_image(batch.frame_tensors[i]))
    if batch.segmentations[i] is not None:
        display(to_pil_image(batch.segmentations[i]))
```
